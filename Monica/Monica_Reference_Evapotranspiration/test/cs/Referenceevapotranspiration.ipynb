```
using System;
using System.Collections.Generic;
using System.Linq;
public class ReferenceEvapotranspiration
{
    private double _height_nn;
    public double height_nn
    {
        get { return this._height_nn; }
        set { this._height_nn= value; } 
    }
    private double _latent_heat_of_vaporization;
    public double latent_heat_of_vaporization
    {
        get { return this._latent_heat_of_vaporization; }
        set { this._latent_heat_of_vaporization= value; } 
    }
    /// <summary>
    /// Constructor of the ReferenceEvapotranspiration component")
    /// </summary>  
    public ReferenceEvapotranspiration() { }
    
    public void  CalculateModel(AMEIRefETState s, AMEIRefETState s1, AMEIRefETRate r, AMEIRefETAuxiliary a, AMEIRefETExogenous ex)
    {
        //- Name: ReferenceEvapotranspiration -Version: 1, -Time step: 1
        //- Description:
    //            * Title: MONICA reference evapotranspiration
    //            * Authors: Claas Nendel (transcription into Crop2ML by Michael Berg-Mohnicke)
    //            * Reference: 
    //            Allen RG, Pereira LS, Raes D, Smith M. (1998) Crop evapotranspiration. Guidelines for computing crop water requirements. FAO Irrigation and
    //            Drainage Paper 56, FAO, Roma
    //        
    //            * Institution: ZALF e.V.
    //            * ExtendedDescription: 
    //            A method following Penman-Monteith as described by the FAO in Allen
    //            RG, Pereira LS, Raes D, Smith M. (1998) Crop evapotranspiration.
    //            Guidelines for computing crop water requirements. FAO Irrigation and
    //            Drainage Paper 56, FAO, Roma
    //        
    //            * ShortDescription: Calculates the MONICA reference evapotranspiration
        //- inputs:
    //            * name: height_nn
    //                          ** description : height above sea level
    //                          ** inputtype : parameter
    //                          ** parametercategory : constant
    //                          ** datatype : DOUBLE
    //                          ** max : 9999
    //                          ** min : -9999
    //                          ** default : 0
    //                          ** unit : m
    //            * name: latent_heat_of_vaporization
    //                          ** description : latent heat of vaporization lambda (0: MONICA's constants, 0.408 = 1/2.45 and psychrometric constant 0.000665 * pressure)
    //                          ** inputtype : parameter
    //                          ** parametercategory : constant
    //                          ** datatype : DOUBLE
    //                          ** max : 10
    //                          ** min : 0
    //                          ** default : 0
    //                          ** unit : MJ/kg
    //            * name: mean_air_temperature
    //                          ** description : daily average air temperature
    //                          ** inputtype : variable
    //                          ** variablecategory : exogenous
    //                          ** datatype : DOUBLE
    //                          ** max : 100
    //                          ** min : -100
    //                          ** default : 0
    //                          ** unit : °C
    //            * name: wind_speed
    //                          ** description : wind speed measured at wind speed height
    //                          ** inputtype : variable
    //                          ** variablecategory : exogenous
    //                          ** datatype : DOUBLE
    //                          ** max : 9999
    //                          ** min : 0
    //                          ** default : 0
    //                          ** unit : m/s
    //            * name: wind_speed_height
    //                          ** description : height at which the wind speed has been measured
    //                          ** inputtype : variable
    //                          ** variablecategory : exogenous
    //                          ** datatype : DOUBLE
    //                          ** max : 9999
    //                          ** min : 0.1
    //                          ** default : 2
    //                          ** unit : m
    //            * name: net_radiation
    //                          ** description : net radiation
    //                          ** inputtype : variable
    //                          ** variablecategory : auxiliary
    //                          ** datatype : DOUBLE
    //                          ** max : 
    //                          ** min : 0
    //                          ** default : 0
    //                          ** unit : MJ/m2
    //            * name: stomata_resistance
    //                          ** description : stomata resistance (FAO default value is 100)
    //                          ** inputtype : variable
    //                          ** variablecategory : auxiliary
    //                          ** datatype : DOUBLE
    //                          ** max : 10000
    //                          ** min : 0
    //                          ** default : 100
    //                          ** unit : s/m
    //            * name: saturation_vapor_pressure_deficit
    //                          ** description : saturation_vapor_pressure_deficit
    //                          ** inputtype : variable
    //                          ** variablecategory : auxiliary
    //                          ** datatype : DOUBLE
    //                          ** max : 
    //                          ** min : 0
    //                          ** default : 0
    //                          ** unit : kPa
    //            * name: saturated_vapor_pressure_slope
    //                          ** description : slope of the saturation vapour pressure curve (negative: calculated from the mean air temperature)
    //                          ** inputtype : variable
    //                          ** variablecategory : auxiliary
    //                          ** datatype : DOUBLE
    //                          ** max : 
    //                          ** min : -1
    //                          ** default : -1
    //                          ** unit : kPa/°C
    //            * name: psychrometric_constant
    //                          ** description : psychrometric constant (negative: calculated from height_nn and latent_heat_of_vaporization)
    //                          ** inputtype : variable
    //                          ** variablecategory : auxiliary
    //                          ** datatype : DOUBLE
    //                          ** max : 
    //                          ** min : -1
    //                          ** default : -1
    //                          ** unit : kPa/°C
        //- outputs:
    //            * name: reference_evapotranspiration
    //                          ** description : reference evapotranspiration (ET0)
    //                          ** variablecategory : auxiliary
    //                          ** datatype : DOUBLE
    //                          ** max : 
    //                          ** min : 0
    //                          ** unit : mm
        double mean_air_temperature = ex.mean_air_temperature;
        double wind_speed = ex.wind_speed;
        double wind_speed_height = ex.wind_speed_height;
        double net_radiation = a.net_radiation;
        double stomata_resistance = a.stomata_resistance;
        double saturation_vapor_pressure_deficit = a.saturation_vapor_pressure_deficit;
        double saturated_vapor_pressure_slope = a.saturated_vapor_pressure_slope;
        double psychrometric_constant = a.psychrometric_constant;
        double reference_evapotranspiration;
        double atmospheric_pressure;
        atmospheric_pressure = 101.3 * Math.Pow((293.0 - (0.0065 * height_nn)) / 293.0, 5.26);
        double psycrometer_constant;
        if (psychrometric_constant >= 0.0)
        {
            psycrometer_constant = psychrometric_constant;
        }
        else if ( latent_heat_of_vaporization > 0.0)
        {
            psycrometer_constant = 1.013e-3 * atmospheric_pressure / (0.622 * latent_heat_of_vaporization);
        }
        else
        {
            psycrometer_constant = 0.000665 * atmospheric_pressure;
        }
        double saturated_vapour_pressure_slope;
        if (saturated_vapor_pressure_slope >= 0.0)
        {
            saturated_vapour_pressure_slope = saturated_vapor_pressure_slope;
        }
        else
        {
            saturated_vapour_pressure_slope = 4098.0 * (0.6108 * Math.Exp(17.27 * mean_air_temperature / (mean_air_temperature + 237.3))) / ((mean_air_temperature + 237.3) * (mean_air_temperature + 237.3));
        }
        double inverse_lambda;
        if (latent_heat_of_vaporization > 0.0)
        {
            inverse_lambda = 1.0 / latent_heat_of_vaporization;
        }
        else
        {
            inverse_lambda = 0.408;
        }
        double wind_speed_2m;
        wind_speed_2m = Math.Max(0.5, wind_speed * (4.87 / Math.Log((67.8 * wind_speed_height - 5.42))));
        double aerodynamic_resistance;
        aerodynamic_resistance = 208.0 / wind_speed_2m;
        double surface_resistance;
        surface_resistance = stomata_resistance / 1.44;
        reference_evapotranspiration = (inverse_lambda * saturated_vapour_pressure_slope * net_radiation + (psycrometer_constant * (900.0 / (mean_air_temperature + 273.0)) * wind_speed_2m * saturation_vapor_pressure_deficit)) / (saturated_vapour_pressure_slope + (psycrometer_constant * (1.0 + (surface_resistance / aerodynamic_resistance))));
        if (reference_evapotranspiration < 0.0)
        {
            reference_evapotranspiration = 0.0;
        }
        a.reference_evapotranspiration= reference_evapotranspiration;
    }
}
```
